```python
from rag import RAG
from openai import OpenAI
import os
from sqlitesearch import VectorSearchIndex
from embedder import Embedder

openai_client = OpenAI(base_url=os.getenv("LLM_BASE_URL"), api_key=os.getenv("LLM_API_KEY"))
embedder = Embedder()

index = VectorSearchIndex(
    mode="ivf",
    db_path="employee_handbook_vectors.db"
)

assistant = RAG(
    embedder=embedder,
    index=index,
    llm_client=openai_client,
)
```

```python
result = assistant.rag("What insurance benefits does the company provide?")
```
